```python
kode = input("Masukan kode tiket:")
match kode:
   case "REG":
     nama="Reguler"
     harga=10000
   case "PREM":
     nama="Premiere"
     harga=20000
   case "VIP":
     nama="VIP"
     harga=30000
print("Nama Tiket:", nama)
print("Harga : ", harga)
```
